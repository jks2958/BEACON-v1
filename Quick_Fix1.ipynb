```python
# %% [markdown]
# BEACON — Quick fix: NaN in coefficient-of-variation columns
# Run this once before moving to Step 11 training.
# Root cause: COV = std / mean, which is undefined (NaN) when mean = 0 —
# happens for degenerate single-packet-style flows with no variation to
# measure. Convention: treat as 0 (no variability to report).

# %%
import pandas as pd

FINAL_DIR = r"D:\Malware Dataset\processed_final"

net_train = pd.read_parquet(FINAL_DIR + r"\net_train.parquet")
net_test = pd.read_parquet(FINAL_DIR + r"\net_test.parquet")

COV_COLS = ["packets_IAT_cov", "cov_packets_delta_time"]

for col in COV_COLS:
    if col in net_train.columns:
        n_fixed = net_train[col].isnull().sum()
        net_train[col] = net_train[col].fillna(0)
        print(f"net_train.{col}: filled {n_fixed} NaN with 0")
    if col in net_test.columns:
        n_fixed = net_test[col].isnull().sum()
        net_test[col] = net_test[col].fillna(0)
        print(f"net_test.{col}: filled {n_fixed} NaN with 0")

net_train.to_parquet(FINAL_DIR + r"\net_train.parquet")
net_test.to_parquet(FINAL_DIR + r"\net_test.parquet")
print("\nSaved. Re-run beacon_validate_before_training.py to confirm [OK] on Check 2.")
```
